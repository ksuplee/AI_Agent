# 📘 프롬프트 엔지니어링 마스터: Zero-shot & Few-shot

이 파일은 **Zero-shot**과 **Few-shot** 프롬프팅의 차이점을 분석하고, 실제 업무 자동화에 어떻게 적용하는지 정리한 가이드입니다.

## 1. Zero-shot 프롬프팅 (직관적 명령)
예시를 제공하지 않고 모델의 사전 학습 지식에만 의존하여 답변을 얻는 방식입니다.

### 📝 실행 시나리오
- **입력 문장:** "서비스는 빠르지만 직원이 불친절했다."
- **프롬프트:** "다음 문장을 긍정 또는 부정으로 분류하시오."
- **AI 응답:** "이 문장은 서비스의 속도는 만족스럽지만 인적 서비스에 불만을 나타내므로 **부정** 혹은 **혼합**으로 볼 수 있습니다."

### ⚠️ 특징
- 별도의 예시를 작성할 필요가 없어 빠르고 간편합니다.
- 하지만 결과의 형식이 모델마다 다를 수 있고, 답변이 길어질 수 있습니다.

## 2. Few-shot 프롬프팅 (패턴 학습)
모델에게 몇 가지 예시를 보여줌으로써 **작업의 기준**과 **출력 형식**을 가르치는 방식입니다.

### 📝 실행 시나리오
**프롬프트:**
1. 문장: 이 음식은 정말 맛있었다. 답변: 긍정
2. 문장: 배송이 너무 늦었다. 답변: 부정
3. 문장: 서비스는 빠르지만 직원이 불친절했다. 답변:

**AI 응답:** **부정**

### ✅ 특징
- 사용자가 원하는 딱 한 단어(긍정/부정)만 출력하도록 강제할 수 있습니다.
- 복잡하거나 모호한 문장의 분류 기준을 예시를 통해 명확히 정립할 수 있습니다.

## 3. 심화 분석: 왜 업무 자동화는 Few-shot인가?

### ❓ Q1. 출력이 더 명확했던 방식은?
**정답: Few-shot**
- 예시를 통해 답변의 구조를 복제하므로 군더더기 없는 결과값을 얻을 수 있습니다.

### ❓ Q2. 예시는 모델에게 무엇을 알려주었는가?
1. **Task Logic (논리):** 복합적인 문장에서 어떤 결론을 내릴지에 대한 판단 기준.
2. **Output Format (형식):** `답변: [결과]`와 같은 일정한 응답 구조.

### ❓ Q3. 실제 업무 자동화에 더 적합한 방식은?
**정답: Few-shot**
- 데이터베이스 저장이나 대량의 텍스트 정제 시 결과값이 일관되어야 오류가 발생하지 않기 때문입니다.

## 4. 최종 비교 테이블

| 항목 | Zero-shot | Few-shot |
| :--- | :--- | :--- |
| **작동 원리** | 모델의 일반 상식 활용 | 사용자가 제공한 패턴 학습 |
| **정확도** | 보통 (가변적) | **높음 (일관적)** |
| **응답 길이** | 서술형 가능성 높음 | **단답/지정 형식 최적화** |
| **권장 용도** | 단순 질문, 브레인스토밍 | **데이터 라벨링, 로그 정제, 자동화** |
