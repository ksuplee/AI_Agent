### 03-2-4 모델 실습

#### (1)  Pipeline으로 텍스트 생성

```python
from transformers import pipeline

generator = pipeline("text-generation", model="gpt2")
generator("AI will change the world because", max_length=30)
```

#### (2) AutoModel + Tokenizer 직접 사용

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

tokenizer = AutoTokenizer.from_pretrained("gpt2")
model = AutoModelForCausalLM.from_pretrained("gpt2")

inputs = tokenizer("AI is", return_tensors="pt")
outputs = model.generate(**inputs, max_length=20)
print(tokenizer.decode(outputs[0]))
```

#### (3) Fine-tuned 감성분석 모델 탐색 및 실행
- 접근 경로: https://huggingface.co/models  
- 검색 필터 활용  
    - Task별: text-classification, summarization, translation, question-answering  
    - Framework별: PyTorch, TensorFlow, JAX  
    - Sort by: downloads, likes, recently updated  
- 예시: 감성분석용 Fine-tuned 모델 탐색  

```python
from transformers import pipeline
classifier = pipeline("text-classification", model="distilbert-base-uncased-finetuned-sst-2-english")
print(classifier("I love this AI course!"))
```

→ 결과: [{'label': 'POSITIVE', 'score': 0.999}]


#### (4) 질의응답 모델 불러오기

```python
from transformers import pipeline

qa = pipeline("question-answering")

context = """AI agents are intelligent systems that can perceive the environment,
make decisions, and take actions autonomously. They are increasingly powered by
large language models (LLMs) such as GPT or LLaMA."""

question = "What powers modern AI agents?"
result = qa(question=question, context=context)
print("질문:", question)
print("답변:", result['answer'])
```
