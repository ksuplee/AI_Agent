### 03-3-3 Text Generation 실습
> Task: 문장을 이어 쓰는 텍스트 생성(Text Generation)

#### (1) GPT-2 기반 텍스트 생성
- 실행 코드  

```python
from transformers import pipeline

# (1) 텍스트 생성 파이프라인 생성
# - task: "text-generation"
# - model: GPT-2 (사전학습된 생성 모델)
generator = pipeline(
    task="text-generation",
    model="gpt2"
)

# (2) 프롬프트(시작 문장) 정의
prompt = "Artificial Intelligence will change"

# (3) 텍스트 생성 실행
# max_length: 생성될 전체 문장 길이
# num_return_sequences: 생성 결과 개수
result = generator(
    prompt,
    max_length=50,
    num_return_sequences=1
)

# (4) 결과 출력
print("=== 생성 결과 ===\n")
print(result[0]["generated_text"])
```

- 실행 후 관찰하기  
    - 문장이 자연스럽게 이어지는가?  
    - 문맥 오류나 반복이 발생하는가?  
    - 프롬프트를 “Artificial Intelligence will change education” 등으로 바꿔보면 결과가 어떻게 달라질까?

### 2. 코드에서 Hugging Face 토큰 사용하기

이제 Colab 보안 저장소에 저장된 `HF_TOKEN`을 코드에서 불러와 Hugging Face에 로그인할 수 있습니다. 아래 코드를 실행해 주세요.

```python
from huggingface_hub import login
from google.colab import userdata

# Colab 보안 저장소에서 HF_TOKEN 불러오기
hf_token = userdata.get('HF_TOKEN')


# Hugging Face 로그인
login(token=hf_token)

print("Hugging Face에 성공적으로 로그인했습니다!")
```

로그인이 성공하면 이제 `Helsinki-NLP/opus-mt-ko-en`와 같은 모델을 문제없이 다운로드하고 사용할 수 있습니다. 번역 코드를 다시 실행해 보세요.

```python
from transformers import pipeline

# (1) 번역 파이프라인 생성
# 한국어 → 영어 번역 모델 지정
translator = pipeline(
    task="translation",
    # model="chunwoolee0/mt5_small_bongsoo_en_ko"
    model="Helsinki-NLP/opus-mt-ko-en"
)

# (2) 번역할 한국어 문장 입력
# text = "AI will change education and society."
text = "인공지능은 교육과 사회를 변화시킬 것입니다."


# (3) 번역 실행
result = translator(
    text,
    max_length=100
)

# (4) 번역 결과 출력
# print("=== 번역 결과 (영어 -> 한국어) ===\n")
print("=== 번역 결과 (한국어 -> 영어) ===\n")
print(result[0]["translation_text"])
```

- 실습 응용  
    - 입력 문장을 바꿔보세요: "AI will transform our daily life in the next decade."  
    - 번역 방향 변경 (한국어 → 영어): "Helsinki-NLP/opus-mt-ko-en" 모델로 변경  
> Tip: Hugging Face Hub에서 “translation” 태그로 더 다양한 언어쌍 모델을 찾을 수 있음  

### 03-3-5 Summarization 실습
- Task: 긴 문장을 요약 (Text Summarization)  

#### (1) 예제 코드: BART 요약 모델
- 실행 코드  



```python
from transformers import pipeline

# (1) 요약 파이프라인 생성
# CNN 뉴스 요약에 Fine-tuned된 BART 모델
summarizer = pipeline(
    task="summarization",
    model="facebook/bart-large-cnn"
)

# (2) 요약할 원문 텍스트
text = """
Artificial intelligence (AI) is rapidly changing the world.
It is being used in education, healthcare, and industry to improve efficiency and accuracy.
However, there are also ethical challenges such as privacy and bias.
"""

# (3) 요약 실행
# max_length: 요약 최대 길이
# min_length: 요약 최소 길이
# do_sample=False: 결과를 안정적으로 생성
result = summarizer(
    text,
    max_length=50,
    min_length=20,
    do_sample=False
)

# (4) 요약 결과 출력
print("=== 요약 결과 ===\n")
print(result[0]["summary_text"])
```

- 실습 응용  
    - 직접 기사나 뉴스 텍스트를 붙여넣어  요약해보기  
    - max_length, min_length 값을 바꿔서 결과 비교  
- 관찰하기  
    - 짧게 요약할수록 정보가 손실되었는가?  
    - 요약 내용이 원문과 일치하는가?
