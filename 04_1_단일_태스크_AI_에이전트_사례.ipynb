##### 3) 코드에서 Hugging Face 토큰 사용하기

이제 Colab 보안 저장소에 저장된 `HF_TOKEN`을 코드에서 불러와 Hugging Face에 로그인할 수 있습니다. 아래 코드를 실행해 주세요.

```python
from huggingface_hub import login
from google.colab import userdata

# Colab 보안 저장소에서 HF_TOKEN 불러오기
hf_token = userdata.get('HF_TOKEN')


# Hugging Face 로그인
login(token=hf_token)

print("Hugging Face에 성공적으로 로그인했습니다!")
```

##### 4) 감성 분석 Fine-tuned 모델 실습

```python
from transformers import pipeline

# 감성 분석용 Fine-tuned 모델 로드
sentiment_agent = pipeline(
    task="text-classification",
    model="distilbert-base-uncased-finetuned-sst-2-english"
)

# 에이전트 입력 (리뷰 텍스트)
review = "The service was fast and the staff was friendly."

# 에이전트 실행
result = sentiment_agent(review)

print(result)
```

#### (2) 실습 2. 요약 에이전트 실행

```python
from transformers import pipeline

# 요약 에이전트용 Fine-tuned 모델
summary_agent = pipeline(
    task="summarization",
    model="facebook/bart-large-cnn"
)

text = """
Artificial intelligence is transforming many industries.
It improves efficiency but also raises ethical concerns.
"""

summary = summary_agent(text, max_length=40, min_length=15, do_sample=False)
print("\n\n--- Summary Result 1 ---\n")
print(summary[0]["summary_text"])
```

- 관찰 포인트  
    - 생성 모델과 비교했을 때 차이는?  
    - 입력 텍스트 < 최대 크기
    - 출력이 예측 가능한가?  

```python
from transformers import pipeline

# 요약 에이전트용 Fine-tuned 모델
summary_agent = pipeline(
    task="summarization",
    model="facebook/bart-large-cnn"
)

text = """
Artificial intelligence is transforming many industries.
It is being widely adopted in fields such as healthcare, finance, manufacturing, and education.
By automating repetitive tasks and analyzing large volumes of data, AI significantly improves efficiency and productivity.
It also enables more accurate decision-making and personalized services for users.
However, the rapid spread of AI raises serious ethical concerns, including issues of privacy, bias, and accountability.
Addressing these challenges is essential to ensure that AI is developed and used responsibly.
"""

summary = summary_agent(text, max_length=40, min_length=15, do_sample=False)
print("\n\n--- Summary Result 2 ---\n")
print(summary[0]["summary_text"])
```

- 관찰 포인트  
    - 입력 텍스트 > 최대 크기
    - 출력이 예측 가능한가?
