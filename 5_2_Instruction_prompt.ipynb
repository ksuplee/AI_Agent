# 🤖 Prompt Engineering Session: Instruction Prompting
이 파일은 AI와 나눈 프롬프트 엔지니어링 학습 및 실습 내용을 포함하고 있습니다.

## 1. 프롬프트의 개념 및 구성 요소
- **정의**: AI에게 지시를 내리는 명령어
- **핵심 요소**: 지시(Instruction), 문맥(Context), 입력 데이터(Data), 출력 형식(Output Format)

## 2. 실습 데이터 (정부 정책 문구)
> "정부는 디지털 전환을 가속화하기 위해 공공 데이터 개방을 확대하고, AI 기술을 활용한 행정 혁신을 추진할 계획이라고 밝혔다."

## 3. Instruction 적용 결과 (정책 분석가 페르소나)
1. 정부는 데이터 민간 개방을 확대함.
2. AI 기술로 행정 효율성을 높임.
3. 시민 체감형 맞춤 서비스 제공이 목적임.
